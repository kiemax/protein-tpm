```python

def find_intersections(dict_of_lists):

    # Convert all lists in the dictionary to sets
    sets_dict = {k: set(v) for k, v in dict_of_lists.items()}

    # Initialize a list to keep track of intersections
    intersections = []

    # Get all the keys of the dictionary
    keys = list(sets_dict.keys())

    # Check for intersections between each pair of sets
    for i in range(len(keys)):
        for j in range(i + 1, len(keys)):
            set1 = sets_dict[keys[i]]
            set2 = sets_dict[keys[j]]
            intersection = set1.intersection(set2)
            if intersection:
                intersections.append((keys[i], keys[j], intersection))

    return intersections
```

```python
def get_holdout_clusters(cluster_type):
    print("Getting holdout clusters")
    pdb_ids = [pdb_id[:4] for pdb_id in clusters[10][0].keys()]
    pdb_chain_identifier = {pdb_id[:4]: pdb_id for pdb_id in clusters[10][0].keys()}  
    holdout_cluster_nums = []
    holdout_cluster = []
    for test_pdb in force_test:
        if not test_pdb.lower() in pdb_ids:
            print("NOT FOUND", test_pdb)
        else:
            print(test_pdb, pdb_chain_identifier[test_pdb.lower()])
            print(clusters[cluster_type][0][pdb_chain_identifier[test_pdb.lower()]])
            holdout_cluster_nums.append(clusters[cluster_type][0][pdb_chain_identifier[test_pdb.lower()]])
    for cluster_num in holdout_cluster_nums:
        holdout_cluster.extend(clusters[cluster_type][3][cluster_num])
    holdout_cluster = list(set(holdout_cluster))
    print("Len holdout cluster", len(holdout_cluster))
    return holdout_cluster
```

```python
def fill_splits(cluster, holdout_cluster):
    # Sizes of the clusters
    cluster_counts = {k:len(v) for k,v in cluster.items()}
    print(cluster_counts)
    cluster_sizes = cluster_counts.values()
    print(cluster_sizes)
    #print(cluster_sizes)
    cluster_tuples = [(l,c) for c,l in cluster_counts.items()]
    # Calculate total size
    total_size = sum(cluster_sizes)+len(holdout_cluster)

    # Calculate target sizes for each split
    train_target = 0.8 * total_size
    val_target = 0.1 * total_size
    test_target = 0.1 * total_size

    print("Targets", train_target, val_target, test_target, "total size", total_size)
    # Initialize the splits
    train_clusters, val_clusters, test_clusters = [], [], holdout_cluster
    train_size, val_size, test_size = 0, 0, 0
    leftover = []

    # Sort clusters by size (largest first) for better fitting
    sorted_clusters = sorted(cluster_tuples, reverse=True)
    print(sorted_clusters)
    # Function to allocate a cluster to the appropriate split
    def allocate_cluster(cluster, target, current_clusters):
        if len(current_clusters) + len(cluster) <= target:
            current_clusters.extend(cluster)
            return current_clusters, True
        return current_clusters, False

    # Allocate clusters to train, validation, and test sets
    for length, cluster_num in sorted_clusters:
        print(length, cluster_num)
        if test_size < test_target:
            test_clusters, success = allocate_cluster(cluster[cluster_num], test_target, test_clusters)
        if not success:
            val_clusters, success = allocate_cluster(cluster[cluster_num], val_target, val_clusters)
        if not success:
            print("Adding to train", len(cluster[cluster_num]))
            train_clusters, success = allocate_cluster(cluster[cluster_num], train_target, train_clusters)
        if not success:
            leftover.append(cluster_num)
            
    # If there are leftover clusters, distribute them to the closest fitting split
    for cluster_num in leftover:
        cluster = cluster[cluster_num]
        if abs(len(train_clusters) + len(cluster) - train_target) < abs(len(val_clusters) + len(cluster) - val_target) and abs(len(train_clusters) + len(cluster) - train_target) < abs(len(test_clusters) + len(cluster) - test_target):
            train_clusters.extend(cluster)
        elif abs(len(val_clusters) + len(cluster) - val_target) < abs(len(test_clusters) + len(cluster) - test_target):
            val_clusters.extend(cluster)
        else:
            test_clusters.extend(cluster)

    # Print the final allocations
    print(f"Train Clusters: , Total Size: {len(train_clusters)} = {len(train_clusters)/total_size} fraction; target {train_target}")
    print(f"Validation Clusters: , Total Size: {len(val_clusters)}= {len(val_clusters)/total_size} fraction; target {val_target}")
    print(f"Test Clusters: , Total Size: {len(test_clusters)}= {len(test_clusters)/total_size} fraction; target {test_target}")
    print(f"Total allocated: {len(train_clusters)+len(val_clusters)+len(test_clusters)} of {total_size}" )
    return train_clusters, val_clusters, test_clusters
```

```python
# actual splitting
traintill_split, valtill_split, testtill_split = fill_splits(cleaned_10, holdout_cluster)
```
