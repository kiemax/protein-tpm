```python
def get_chains(pdbPath):
    #extract chains
    try:
        chain_pairs = extract_chains_chothia(pdbPath)
        if not chain_pairs:
            raise ValueError(f"No chain pairs returned for {pdbPath}")
        ligand_chains, receptor_chains = chain_pairs[0]


        ### Some chains are annotated with int like "2" in sabdab. This leads to unphysical brooms downstream
        if not all(isinstance(v, str)and  v.isalpha() for v in ligand_chains):
            raise ValueError(f"Invalid characters in ligand chains: {ligand_chains}")

        if not all(isinstance(v, str)and  v.isalpha() for v in receptor_chains):
            raise ValueError(f"Invalid characters in receptor chains: {receptor_chains}")

        receptor_chains = fix_case_if_reused(ligand_chains, receptor_chains, Path(pdbPath).stem) # some chothia pdbs have wrong headers

    except Exception as e:
        return None
    return ligand_chains, receptor_chains
```

```python
def extract_chain_ids_pocket(filepath):
    chain_ids = set()
    with filepath.open('r') as file:
        for line in file:
            if line.startswith("ATOM"):
                parts = line.split()
                try:
                    chain_id = parts[4]
                    chain_ids.add(chain_id)
                except IndexError:
                    continue
    return chain_ids
```

```python
def present_in_all_lists(s, tuple_of_lists):
    return all(any(item in lst for item in s) for lst in tuple_of_lists)
```

```python
def filter_files_by_chains(folder_path, target_chains):
    pockets_folder = Path(folder_path)
    #assert pockets_folder.is_dir()
    matching_files = []
    print("right before")
    for pocket_pdb in pockets_folder.glob("*.pdb"):
        print("in the loop")
        chains = extract_chain_ids_pocket(pocket_pdb)
        print(f"chains: {chains}")
        #print(f"target_chains: {target_chains}")
        if present_in_all_lists(chains, target_chains):  # intersection
            matching_files.append(pocket_pdb.name)
    return matching_files
```
