```python
import numpy as np
```

```python
shards_all = ["Carboxamid_start_atm",
               "Guanidino",
               "Hydroxyl",
               "Terminal_amino",
               "Carboxyl",
               "Thiol",
               "Thioether",
               "His",
               "Phe",
               "Pro",
               "Trp",
               "Tyr"]
```

```python
def load_shards(shard_list):

    ### chart_list: list with paths to shard data
    target_positions = []
    target_classes = []

    for shard in shard_list:
        target_position = np.load(f"{shard}/target_positions_train.npz", allow_pickle=True)["arr"][:, : , :]
        target_positions.append(target_position)
        target_position = np.load(f"{shard}/target_positions_val.npz", allow_pickle=True)["arr"][:, : , :]
        target_positions.append(target_position)

        target_class = np.load(f"{shard}/target_classes_train.npz", allow_pickle=True)["arr"][:, : , :]
        target_classes.append(target_class)
        target_class = np.load(f"{shard}/target_classes_val.npz", allow_pickle=True)["arr"][:, : , :]
        target_classes.append(target_class)

    return (np.concatenate(target_positions, axis=0),
            np.concatenate(target_classes, axis=0))
```

```python
shard_data = [load_shards([f"../data/sabdab/interfaces/{shard}"]) for shard in shards_all]
```

```python
for data, shard in zip(shard_data, shards_all):

    np.savez_compressed(f'../data/sabdab/interfaces/{shard}/target_positions_all.npz', arr=data[0])
    np.savez_compressed(f'../data/sabdab/interfaces/{shard}/target_classes_all.npz', arr=data[1])
```
